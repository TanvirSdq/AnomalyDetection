```python
import pandas as pd
import numpy as np

# Make our experiment reproducible
np.random.seed(42)

# Generate 100 fictional heart-rate readings
heart_rates = np.random.normal(
    loc=70,
    scale=5,
    size=100
)

# Put the readings into a table
df = pd.DataFrame({
    "heart_rate": heart_rates
})

print(df.head())
print("\nTotal readings:", len(df))
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Make the experiment repeatable
np.random.seed(42)

# Generate 200 fictional heart-rate readings
df = pd.DataFrame({
    "heart_rate": np.random.normal(
        loc=70,
        scale=5,
        size=200
    )
})

print(df.head(10))
print("\nSummary:")
print(df["heart_rate"].describe())
```

```python

plt.figure(figsize=(10, 4))

plt.plot(df["heart_rate"], marker=".", linestyle="-")
plt.axhline(
    df["heart_rate"].mean(),
    linestyle="--",
    label="Average"
)

plt.title("Fictional Astronaut Heart-Rate Readings")
plt.xlabel("Observation number")
plt.ylabel("Heart rate (BPM)")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python

# Learn the baseline from historical readings
mean = df["heart_rate"].mean()
std = df["heart_rate"].std()

# Define a simple illustrative threshold
lower_limit = mean - 3 * std
upper_limit = mean + 3 * std

# Flag readings outside the limits
df["is_anomaly"] = (
    (df["heart_rate"] < lower_limit) |
    (df["heart_rate"] > upper_limit)
)

print("Learned average:", round(mean, 2))
print("Learned standard deviation:", round(std, 2))
print("Flagged readings:", df["is_anomaly"].sum())
print(df[df["is_anomaly"]])
```
